```python
import os 
from dotenv import load_dotenv
import smtplib
import mimetypes
from email.message import EmailMessage

load_dotenv()

My_email = 'rihamhussain744@gmail.com'
my_pass = os.getenv("app_pass")
Receiver = input("Enter email of Reciever:")
subject = input("Enter Subject:")

msg = EmailMessage()
msg['Subject'] = subject
msg['From'] = My_email
msg['To'] = Receiver

print('\nEnter your email content (press Enter on an empty line to finish):')
boby_content = []
while True:
    line = input()
    if line =="":
        break
    boby_content.append(line)

final_body = "\n".join(boby_content)
msg.set_content(final_body)

print(" Type a file path and press Enter. To FINISH on empty line.")

while True:
    file_path = input("Attach file: ").strip()
    if not file_path:
        break

    try:
        with open(file_path, 'rb') as f:
            file_data = f.read()
            file_name = f.name

            ctype, encoding = mimetypes.guess_type(file_path)
            if ctype is None or encoding is not None:
                ctype = 'application/octet-stream'
            maintype, subtype = ctype.split('/', 1)

            msg.add_attachmet(
                file_data,
                maintype=maintype,
                subtype=subtype,
                filename=file_path
            )
        print(f"Added: {file_name}")
    except FileNotFoundError:
        print(f"Error: '{file_path}' not found. try again.")

try:
    with smtplib.SMTP_SSL('smtp.gmail.com', 465) as smtp:
        smtp.login(My_email, my_pass)
        smtp.send_message(msg)
    print("\n Email sent successfully with all available attachments!")
except Exception as e:
    print(f"Error: {e}")
```
